```python
#8 acccept a no and check if it is a perfect no or not,(perfect no- whose sum of factors is equal to the number itself)
#ex - 6=1,2,3= sum is 6 itself
n = int(input("please put your number to check if its a perfect number-"))
sum = 0   #naya variable to store an retuen sum of th efactors

for i in range(1,n):      #1 se leke n tk liya, mtlb 6 h toh 6 k pehle 5 tk aayega kyuki 6 cannto be included as a factor
    if n%i == 0:          #1,2,3 factors honge 6 ke toh remainder 0
        sum = sum + i      #sum me sum + iterative saare 1 se 5 tk k factors aayenge ad hoke
                        #zaruri nahi if k niche else aana vchahiye
if sum == n:
    print(f"{n} is a perfect number with sum {sum}")
else:
    print(f"{n} is not a perfect numner")
        
```

```python
#coutn letters, digits and special symbols from a given string, eg str="uyasfd7333#@$$", output- char=, digits=a and symbol=
#search string methods python on chroem and check https://www.w3schools.com/python/python_ref_string.asp, different functions of stings
# got to knoe ki isdigit(), isaplha(), give output true or false only if in the string there is num or characters

a = "427324hfgfskjjf@#@^#@&%#"

dig = 0 #digits to be counted start me coout 0 
char = 0 
spchr = 0 #specidal characters to be counted
for i in a: #accesing dircetly the character of the string a
    if i.isdigit():
        dig += 1 #if there digits in str then it would increase outn,dig += 1, is same as dig = dig +1
    elif i.isalpha():
        char += 1
    else:
        spchr += 1
print(f"digits={dig}\ncharactes={char}\nspecial caracters={spchr}")
    
```
