```python
#questions on while loop
#1 take input of numner and separate each digit of the number and print on a new line

# =====================================================================
# LINE-BY-LINE EXPLANATION: EXTRACTING & PRINTING DIGITS
# Example Number Used: 125
# =====================================================================

# --- 1. THE SETUP ---

# input() takes the typed number as a string (e.g., "125").
# int() converts that string into a mathematical integer (n = 125).
n = int(input("please give your number to separate="))


# --- 2. THE LOOP ---

# The loop keeps running as long as 'n' is greater than 0.
while n > 0:
    
    # n % 10 calculates the remainder when divided by 10 (the last digit).
    # Since it is inside a print() statement, it immediately prints that digit 
    # to the screen and moves to a new line.
    # Example flow:
    # Loop 1: prints 5
    # Loop 2: prints 2
    # Loop 3: prints 1
    print(n % 10) 
    
    # n // 10 performs floor division to remove the last digit from 'n'.
    # This prepares 'n' for the next cycle of the loop.
    # Example flow:
    # Loop 1: 125 // 10 = 12
    # Loop 2: 12 // 10 = 1
    # Loop 3: 1 // 10 = 0 (This makes n=0, ending the loop)
    n = n // 10

# Note: There is no final print statement outside the loop here, 
# because the digits are being printed one by one inside the loop!
# =====================================================================
# 🗺️ STEP-BY-STEP EXECUTION TRACE (Input: 125)
# =====================================================================
# This is a "dry run" of the program flow, showing exactly how Python 
# navigates the code and updates variables in memory line-by-line.

# --- 1. SEQUENTIAL SETUP ---
# Code:   n = int(input(...))
#         print(...)
# Action: Prompts user and converts input to integer, then prints the header text.
# State:  n = 125
# Output: --- Output using Math...


# --- 2. ITERATIVE CHECK 1 ---
# Code:   while n > 0:
# Action: Evaluates if 125 > 0. The result is True!
# State:  Python approves entry into the loop block.


# --- 3. LOOP CYCLE 1 ---
# Code:   print(n % 10)
#         n = n // 10
# Action: Calculates 125 % 10 (which is 5) and prints it.
#         Calculates 125 // 10 (which is 12) and updates n.
# State:  n is updated to 12.
# Output: 5


# --- 4. ITERATIVE CHECK 2 ---
# Code:   while n > 0:
# Action: Jumps back to the top condition. Evaluates if 12 > 0 (True).
# State:  Python approves entry into the loop block again.


# --- 5. LOOP CYCLE 2 ---
# Code:   print(n % 10)
#         n = n // 10
# Action: Calculates 12 % 10 (which is 2) and prints it.
#         Calculates 12 // 10 (which is 1) and updates n.
# State:  n is updated to 1.
# Output: 2


# --- 6. ITERATIVE CHECK 3 ---
# Code:   while n > 0:
# Action: Jumps back to the top condition. Evaluates if 1 > 0 (True).
# State:  Python approves entry into the loop block again.


# --- 7. LOOP CYCLE 3 ---
# Code:   print(n % 10)
#         n = n // 10
# Action: Calculates 1 % 10 (which is 1) and prints it.
#         Calculates 1 // 10 (which is 0) and updates n.
# State:  n is updated to 0.
# Output: 1


# --- 8. LOOP EXIT ---
# Code:   while n > 0:
# Action: Jumps back to the top condition. Evaluates if 0 > 0 (FALSE).
# State:  Python destroys the loop and skips the block entirely. 
#         The loop is finished!
```

```python
#2 accept a numner and print its reverse
# =====================================================================
# LINE-BY-LINE EXPLANATION: REVERSING A NUMBER
# Example Number Used: 251
# =====================================================================

# --- 1. THE SETUP ---

# input() takes the user's typed number as a string (e.g., "251").
# int() converts that string into a mathematical integer (n = 251).
n = int(input("please give your number to separate="))

# We create a variable 'rev' starting at 0 to build our reversed number.
rev = 0


# --- 2. THE LOOP ---

# The loop runs repeatedly as long as 'n' is greater than 0.
while n > 0:
    
    # n % 10 extracts the LAST digit (e.g., 251 % 10 = 1).
    # rev * 10 shifts the current reversed digits left to make room.
    # Adding them together appends the new digit. 
    # Example flow:
    # Loop 1: (0 * 10) + 1 = 1
    # Loop 2: (1 * 10) + 5 = 15
    # Loop 3: (15 * 10) + 2 = 152
    rev = rev * 10 + n % 10
    
    # n // 10 removes the last digit from 'n' using floor division.
    # Example flow:
    # Loop 1: 251 // 10 = 25
    # Loop 2: 25 // 10 = 2
    # Loop 3: 2 // 10 = 0 (This makes n=0, causing the loop to stop next time)
    n = n // 10


# --- 3. THE FINISH LINE ---

# Once n reaches 0, the loop breaks. The final reversed number is printed.
print(rev)
   
# =====================================================================
# 🗺️ STEP-BY-STEP EXECUTION TRACE (Input: 251)
# =====================================================================
# This "dry run" shows exactly how Python builds the reversed number
# in memory, step-by-step.

# --- 1. SEQUENTIAL SETUP ---
# Code:   n = int(input(...))
#         rev = 0
# Action: Prompts user and sets up the variables.
# State:  n = 251
#         rev = 0


# --- 2. ITERATIVE CHECK 1 ---
# Code:   while n > 0:
# Action: Evaluates if 251 > 0. (True)
# State:  Python approves entry into the loop block.


# --- 3. LOOP CYCLE 1 ---
# Code:   rev = rev * 10 + n % 10
#         n = n // 10
# Action: Calculates (0 * 10) + (251 % 10)  --> 0 + 1 = 1
#         Calculates 251 // 10 --> 25
# State:  rev is updated to 1
#         n is updated to 25


# --- 4. ITERATIVE CHECK 2 ---
# Code:   while n > 0:
# Action: Jumps back to the top condition. Evaluates if 25 > 0. (True)
# State:  Python approves entry into the loop block again.


# --- 5. LOOP CYCLE 2 ---
# Code:   rev = rev * 10 + n % 10
#         n = n // 10
# Action: Calculates (1 * 10) + (25 % 10)  --> 10 + 5 = 15
#         Calculates 25 // 10 --> 2
# State:  rev is updated to 15
#         n is updated to 2


# --- 6. ITERATIVE CHECK 3 ---
# Code:   while n > 0:
# Action: Jumps back to the top condition. Evaluates if 2 > 0. (True)
# State:  Python approves entry into the loop block again.


# --- 7. LOOP CYCLE 3 ---
# Code:   rev = rev * 10 + n % 10
#         n = n // 10
# Action: Calculates (15 * 10) + (2 % 10)  --> 150 + 2 = 152
#         Calculates 2 // 10 --> 0
# State:  rev is updated to 152
#         n is updated to 0


# --- 8. LOOP EXIT ---
# Code:   while n > 0:
# Action: Jumps back to the top condition. Evaluates if 0 > 0. (FALSE)
# State:  Python destroys the loop and exits.


# --- 9. THE FINISH LINE ---
# Code:   print(rev)
# Action: Prints the final value stored in the 'rev' variable.
# Output: 152
```

```python
#3 check if given int is palindrome or not

# =====================================================================
# LINE-BY-LINE EXPLANATION: PALINDROME CHECKER
# Example Number Used: 121
# =====================================================================

# --- 1. THE SETUP ---

# input() takes the typed number, and int() converts it to an integer.
n = int(input("please give your number to check if palindrome-"))

# CRITICAL STEP: We make a backup of 'n' and store it in 'copy'.
# Why? Because the while loop below will reduce 'n' down to 0. 
# If we don't save the original number now, we'll have nothing to compare 
# our reversed number to at the end! (copy = 121)
copy = n

# We start our reversed number builder at 0.
rev = 0


# --- 2. THE REVERSING LOOP ---

# This is the exact same logic you used before to reverse a number.
while n > 0:
    # Shift current rev left by multiplying by 10, then add the last digit of n.
    rev = rev * 10 + n % 10 
    
    # Remove the last digit from n.
    n = n // 10

# By the end of this loop:
# 'rev' has successfully been built into 121.
# 'n' has been reduced to 0.


# --- 3. THE COMPARISON ---

# The computer checks if our newly built 'rev' (121) exactly matches 
# the original number we backed up in 'copy' (121).
if rev == copy:
    # If they match, it's a palindrome! 
    # The 'f' inside print() allows us to inject the 'rev' variable directly into the text.
    print(f"{rev} is a palindrome")
else:
    # If they don't match (e.g., original was 123, reversed is 321), this runs.
    print("it is not a palindrome")
```

```python
#3 find the greatest elementa nd print its inde too
# =====================================================================
# LINE-BY-LINE EXPLANATION: FINDING THE LARGEST NUMBER & ITS INDEX
# =====================================================================

# --- 1. INITIAL SETUP ---

# We create our list of numbers and store it in the variable 'l'.
l = [12, 567, 43, 235, 347, 568, 45, 7]

# We assume the very first number (at index 0) is the largest one so far.
# largest = 12
largest = l[0]

# We record the position (index) of our current largest number.
# index = 0
index = 0


# --- 2. THE SEARCH LOOP ---

# len(l) gets the total number of items in the list (which is 8).
# range(8) creates a sequence of numbers from 0 to 7.
# 'i' will represent the current index we are looking at during each loop.
for i in range(len(l)):
    
    # We check if the current number we are looking at (l[i]) 
    # is bigger than the biggest number we've found so far ('largest').
    if l[i] > largest:
        
        # If it IS bigger, we update our 'largest' variable with this new number.
        largest = l[i]
        
        # We also update 'index' to remember where we found this new champion.
        index = i


# --- 3. THE RESULT ---

# After checking every single number, the loop finishes.
# We print the final 'largest' number and its 'index' using an f-string.
print(f"your largest number is {largest} at index {index}")
```

```python
#4 find the second greatest largest number in the list
# =====================================================================
# LINE-BY-LINE EXPLANATION: FINDING LARGEST & SECOND LARGEST
# =====================================================================

# --- 1. INITIAL SETUP ---

# We create our list of numbers.
l = [12, 16, 13, 19, 17]

# We assume the very first number is both the largest and second largest to start.
# largest = 12
largest = l[0]

# sec_largest = 12
sec_largest = l[0]


# --- 2. THE SEARCH LOOP ---

# Unlike the previous example that used range(len()), this loop iterates 
# directly over the VALUES in the list. 'i' will be 12, then 16, then 13, etc.
for i in l:
    
    # CONDITION 1: Is the current number 'i' a new all-time high?
    if i > largest:
        
        # If we found a new largest number, the OLD largest number gets demoted.
        # It drops down to become the second largest.
        sec_largest = largest
        
        # Then, we crown the current number 'i' as the new largest.
        largest = i
        
    # CONDITION 2: It wasn't the largest. But is it bigger than our current second largest?
    # (The 'elif' ensures we only check this if Condition 1 was False)
    elif i > sec_largest:
        
        # If it's not a new gold medalist, but it beats the silver medalist,
        # we update the second largest number.
        sec_largest = i


# --- 3. THE RESULT ---

# The loop finishes checking every number.
# We print the silver medalist and the gold medalist.
print(sec_largest, largest)
```

```python
#5 chek if list is sorted or not
# =====================================================================
# LINE-BY-LINE EXPLANATION: CHECKING IF A LIST IS SORTED
# =====================================================================

# --- 1. INITIAL SETUP ---

# We create our list of numbers. 
a = [12, 13, 18, 15, 16]


# --- 2. THE COMPARISON LOOP ---

# len(a) is 5. We subtract 1 to get 4. range(4) means 'i' will be 0, 1, 2, 3.
# Why -1? Because we compare a[i] to the NEXT item a[i+1]. 
# If we went to the very last index, a[i+1] would cause an error!
for i in range(len(a)-1):
    
    # We check if the current number is smaller than the NEXT number.
    if a[i] < a[i+1]:
        # If it is, the list is still sorted so far. 
        # 'continue' tells the loop to immediately jump to the next 'i'.
        continue
        
    # If the current number is NOT smaller than the next (meaning it's bigger or equal),
    # it means the list is out of order.
    else:
        # We print that the list is not sorted.
        print("your list is not sorted")
        
        # 'break' completely stops and exits the loop immediately. 
        # We found an out-of-order pair, so there is no need to check the rest!
        break

# --- 3. THE FOR-ELSE BLOCK ---

# This 'else' belongs to the 'for' loop, NOT the 'if' statement!
# In Python, a for-else block only runs if the loop finished naturally 
# WITHOUT ever hitting a 'break' statement.
else:
    # If the loop finished checking every pair and never hit 'break',
    # it means every number was smaller than the next one.
    print("your list is sorted")
```

```python
#SETS IN PYTHON,set is an unordered collection of unique elements. Sets are primarily used to eliminate duplicate values and perform highly efficient membership testing (checking if an item exists).
# =====================================================================
# 🛡️ SET POWERS: KEY TERMINOLOGY & SOLVED EXAMPLES
# =====================================================================
# Before working with Sets, we need to understand some of their core 
# characteristics. Here is the breakdown along with code examples.

# =====================================================================
# 1. MUTABLE
# =====================================================================
# Sets are mutable, meaning you can change the values of a set by 
# adding or removing elements after it is created.

my_set = {10, 20, 30}

# We can easily add a new element:
my_set.add(40)
print(my_set)
# OUTPUT: {40, 10, 20, 30} (Order may vary!)

# We can also remove elements:
my_set.remove(20)
print(my_set)
# OUTPUT: {40, 10, 30}


# =====================================================================
# 2. DUPLICATES (Not Allowed)
# =====================================================================
# You cannot have any duplicate values in a set. That means every 
# element will be strictly unique. 

# If you try to create a set with duplicates, Python automatically 
# and silently filters them out!
unique_set = {5, 1, 5, 9, 5, 5}

print(unique_set)
# OUTPUT: {1, 5, 9}  <-- Notice all the extra 5s are gone!

# Pro-Tip: This makes Sets the fastest way to remove duplicates from a List!


# =====================================================================
# 3. UNORDERED
# =====================================================================
# Sets are unordered, which means they do not keep track of where items 
# are positioned. Because of this, you CANNOT access them through index values.

unordered_set = {"apple", "banana", "cherry"}

# If we try to grab the first item at index 0 like we do with lists/tuples:
# print(unordered_set[0])

# OUTPUT:
# TypeError: 'set' object is not subscriptable


# =====================================================================
# 4. SEMI-HETEROGENOUS
# =====================================================================
# Sets are semi-heterogenous. They can store different data types 
# (like strings, numbers, and tuples), but NOT everything. 
# Specifically, elements inside a set must be "immutable" (unchangeable).

# This works perfectly (Integer, String, Tuple):
mixed_set = {42, "Python", (1, 2, 3)}
print(mixed_set)
# OUTPUT: {42, 'Python', (1, 2, 3)}

# But if we try to put a mutable object like a List inside a Set, it crashes:
# bad_set = {1, 2, [3, 4]}  

# OUTPUT:
# TypeError: unhashable type: 'list'
```

```python
# =====================================================================
# 🧠 HOW SETS STORE VALUES IN PYTHON (HASHING)
# =====================================================================
# Sets in Python are incredibly fast at finding out if an item exists. 
# This speed comes from a concept called "Hashing." Here is exactly 
# how it works under the hood.

# =====================================================================
# 1. THE HASH FUNCTION
# =====================================================================
# Each value you put in a set is passed through Python's built-in 
# hash() function, which converts the data into a unique integer.

word_hash = hash("apple")
print(f"The hash value for 'apple' is: {word_hash}")
# OUTPUT: (A large, random-looking integer, e.g., 873409831...)


# =====================================================================
# 2. HASH AS A MEMORY INDEX
# =====================================================================
# Python uses that generated hash number as an exact "address" 
# to store the element in the computer's memory. 
#
# When you ask Python if "apple" is in the set, it just calculates 
# the hash again and jumps straight to that memory address! It doesn't 
# have to search through the other items.


# =====================================================================
# 3. WHY SETS ARE UNORDERED
# =====================================================================
# Because items are placed in memory based on their random-looking hash 
# address (and not the order you typed them in), hashing does not 
# maintain order. Therefore, sets are strictly unordered.

my_set = {"banana", "apple", "cherry", "date"}
print(my_set)
# OUTPUT: {'date', 'cherry', 'banana', 'apple'} <-- Order is scrambled!


# =====================================================================
# 4. ONLY IMMUTABLE (HASHABLE) OBJECTS ALLOWED
# =====================================================================
# If an object can change (mutable), its hash would also have to change. 
# Because that would break the memory indexing system, Python only 
# allows immutable (hashable) objects inside sets.

# ✅ Allowed: Numbers, Strings, Tuples (Because they cannot change)
valid_set = {1, "Hello", (1, 2, 3)}
print("Valid set created successfully!")

# ❌ NOT Allowed: Lists, Dictionaries (Because they are mutable)
# invalid_set = {1, 2, [3, 4]}  

# OUTPUT:
# TypeError: unhashable type: 'list'
```

```python
# =====================================================================
# 🚶‍♂️ SET TRAVERSING: COMPLETE GUIDE & EXAMPLES
# =====================================================================
# Traversing means moving through the elements of a data structure 
# one by one. Here is how it works with Sets.

# =====================================================================
# 1. THE PROBLEM WITH INDEXING
# =====================================================================
# A set CANNOT be traversed using index values because it is 
# unordered and has no index. 

my_set = {"apple", "banana", "cherry"}

# If we try to use a standard index-based loop like we do with lists:
# for i in range(len(my_set)):
#     print(my_set[i])

# OUTPUT:
# TypeError: 'set' object is not subscriptable


# =====================================================================
# 2. THE CORRECT WAY TO TRAVERSE (DIRECT ITERATION)
# =====================================================================
# To traverse a set, you must loop through the items directly. 
# Keep in mind: because sets are unordered via memory hashing, traversing 
# them will often give you the values in a seemingly "random" order!

print("--- Basic String Set ---")
for item in my_set:
    print(item)


# =====================================================================
# EXAMPLE A: TRAVERSING A SET OF INTEGERS (Ref: image_ae49e3.png)
# =====================================================================
# Here is a set containing only numbers.

a = {1, 8, 9, 2, 3, 4, 5}

print("\n--- Output for Integer Set ---")
for i in a:
    print(i)
    
# NOTE: While small integers sometimes print in numerical order due to how 
# Python optimizes integer hashes under the hood, you should NEVER rely 
# on this. Sets are strictly unordered.


# =====================================================================
# EXAMPLE B: TRAVERSING A HETEROGENEOUS SET (Ref: image_ae49df.png)
# =====================================================================
# Let's add a string to the mix to see the unordered nature more clearly.

b = {1, 8, 9, "hello", 2, 3, 4, 5}

print("\n--- Output for Mixed Set ---")
for i in b:
    print(i)

# NOTE: Now that we have mixed data types, the hashing mechanism will 
# scramble the output order noticeably. The string "hello" will likely 
# appear in a seemingly random spot when printed!
```

```python
# =====================================================================
# 🛠️ BASIC SET METHODS: ADDING & REMOVING
# =====================================================================
# Sets are mutable, but because they don't have indexing, we must use 
# these specific methods to modify their contents.

# --- INITIAL SETUP ---
s = {1, 2, 3}
print(f"Initial set: {s}")


# =====================================================================
# 1. ADDING ELEMENTS
# =====================================================================
# add() inserts a single new element into the set.
# (If you try to add a duplicate, it just silently ignores it!)

s.add(4)
print(f"After add(4): {s}")
# OUTPUT: {1, 2, 3, 4}


# =====================================================================
# 2. REMOVING ELEMENTS (remove vs. discard)
# =====================================================================
# There are two ways to specify an item to remove, and the difference 
# between them is very important!

# --- remove() ---
# Removes a specific element. 
# WARNING: If the element is NOT in the set, it will crash your program (KeyError).
s.remove(2)
print(f"After remove(2): {s}")
# OUTPUT: {1, 3, 4}

# --- discard() ---
# Removes a specific element, but is "safe". 
# If the element is NOT in the set, it does absolutely nothing and prevents errors.
s.discard(5) 
print(f"After discard(5): {s}")
# OUTPUT: {1, 3, 4} (Notice how it didn't crash, even though 5 wasn't there!)


# =====================================================================
# 3. POPPING & CLEARING
# =====================================================================

# --- pop() ---
# Removes and returns an element. Because sets are unordered, you don't 
# know which element will be popped! It is essentially a "random" element.
popped_element = s.pop()
print(f"Popped element: {popped_element}")
print(f"After pop(): {s}")
# OUTPUT: (Will vary, e.g., Popped element: 1, Remaining: {3, 4})

# --- clear() ---
# Wipes the entire set clean, leaving an empty set.
s.clear()
print(f"After clear(): {s}")
# OUTPUT: set()  <-- This is how Python represents a completely empty set 
```

```python
# =====================================================================
# 🔗 SPECIAL SET OPERATIONS (MATHEMATICAL COMPARISONS)
# =====================================================================
# These operations allow you to compare two sets to find overlapping
# data, unique data, or combine them entirely. 

# --- INITIAL SETUP ---
A = {1, 2, 3}
B = {3, 4, 5}

print(f"Set A: {A}")
print(f"Set B: {B}\n")


# =====================================================================
# 1. UNION (Combine Everything)
# =====================================================================
# Combines all elements from both sets into one large set. 
# Because it's a set, any duplicates (like the '3') are only kept once!

union_set = A.union(B)
# Shortcut operator: A | B

print(f"Union (A | B): {union_set}")
# OUTPUT: {1, 2, 3, 4, 5}


# =====================================================================
# 2. INTERSECTION (Only the Overlap)
# =====================================================================
# Returns ONLY the elements that exist in BOTH Set A and Set B.

intersection_set = A.intersection(B)
# Shortcut operator: A & B

print(f"Intersection (A & B): {intersection_set}")
# OUTPUT: {3}


# =====================================================================
# 3. DIFFERENCE (Subtracting Sets)
# =====================================================================
# Returns elements that are in Set A, but completely removes anything 
# that is also found in Set B.

difference_set = A.difference(B)
# Shortcut operator: A - B

print(f"Difference (A - B): {difference_set}")
# OUTPUT: {1, 2}


# =====================================================================
# 4. SYMMETRIC DIFFERENCE (The Anti-Intersection)
# =====================================================================
# Returns elements that are in EXACTLY ONE of the sets. 
# It completely removes the overlapping items (the '3').

symmetric_diff = A.symmetric_difference(B)
# Shortcut operator: A ^ B

print(f"Symmetric Difference (A ^ B): {symmetric_diff}")
# OUTPUT: {1, 2, 4, 5}


# =====================================================================
# 📝 A NOTE FROM THE SLIDE
# =====================================================================
# The slide mentions: "At the end set is not used that much in python". 
# While they might not be used as constantly as Lists or Dictionaries 
# in basic scripts, Sets are actually an incredible tool whenever you 
# need to remove duplicates from a list instantly or do fast comparisons!
```

```python
#DICTIONARY IN PYTHON-dictionary is a built-in data structure that stores data in key:value pairs, working like a real-world map or address book where you look up information using a unique identifier
# =====================================================================
# 📚 WHAT IS A DICTIONARY? (DEFINITION, SYNTAX & FORMAT)
# =====================================================================
# DEFINITION:
# A dictionary in Python is a built-in data structure used to store 
# data in "key-value" pairs. Think of it like a real-world dictionary: 
# you look up a word (the "key") to instantly find its meaning (the "value").

# SYNTAX & FORMAT:
# - Dictionaries are enclosed in curly brackets {}.
# - Each item is a pair made up of a key and a value.
# - The key and value are separated by a colon (:).
# - Each key-value pair is separated by a comma (,).

# --- SYNTAX TEMPLATE ---
# dictionary_name = {
#     "key1": "value1",
#     "key2": "value2"
# }

# --- BASIC EXAMPLE ---
student_profile = {
    "name": "John Doe",
    "major": "Computer Science",
    "graduation_year": 2025
}

print(f"Basic Dictionary Format: {student_profile}\n")


# =====================================================================
# 📖 DICTIONARY POWERS: KEY TERMINOLOGY & SOLVED EXAMPLES
# =====================================================================
# Before working with Dictionaries, we need to understand some of their 
# core characteristics. Here is the breakdown along with code examples.

# =====================================================================
# 1. MUTABLE
# =====================================================================
# Dictionaries are mutable, meaning you can change, add, or remove 
# key-value pairs after creation.

my_dict = {"name": "Alice", "age": 25}

# Changing an existing value
my_dict["age"] = 26

# Adding a brand new key-value pair
my_dict["city"] = "Mumbai"

print(f"Mutable Dictionary: {my_dict}")
# OUTPUT: Mutable Dictionary: {'name': 'Alice', 'age': 26, 'city': 'Mumbai'}


# =====================================================================
# 2. DUPLICATES (Keys vs. Values)
# =====================================================================
# Keys MUST be unique, but you can have duplicates in values.

# If you try to use the same key twice, the new value overwrites the old one!
duplicate_keys_dict = {"color": "red", "size": "M", "color": "blue"}

print(f"Duplicate Keys (Overwritten): {duplicate_keys_dict}")
# OUTPUT: {'color': 'blue', 'size': 'M'}  <-- "red" is gone!

# However, duplicating VALUES across different keys is perfectly fine:
duplicate_values_dict = {"fruit1": "apple", "fruit2": "apple"}

print(f"Duplicate Values (Allowed): {duplicate_values_dict}")
# OUTPUT: {'fruit1': 'apple', 'fruit2': 'apple'}


# =====================================================================
# 3. ORDER
# =====================================================================
# Dictionaries follow insertion order (as of modern Python 3.7+). 
# This means the order you add items is the order they will stay in.

order_dict = {}
order_dict["first"] = 1
order_dict["second"] = 2
order_dict["third"] = 3

print(f"Insertion Order: {order_dict}")
# OUTPUT: Insertion Order: {'first': 1, 'second': 2, 'third': 3}


# =====================================================================
# 4. HETEROGENEOUS
# =====================================================================
# A dictionary can store different types of keys and values, like 
# integers, strings, lists, or even another dictionary!

mixed_dict = {
    1: "Integer Key",                # Key is int, Value is string
    "fruits": ["apple", "mango"],    # Key is string, Value is a List
    "nested_dict": {"x": 10, "y": 5} # Key is string, Value is a Dictionary
}

print(f"Heterogeneous Dictionary: {mixed_dict}")
# OUTPUT: {1: 'Integer Key', 'fruits': ['apple', 'mango'], 'nested_dict': {'x': 10, 'y': 5}} 
```

```python
# =====================================================================
# ⚙️ DICTIONARY SYNTAX AND WORKING
# =====================================================================
# Dictionaries use "key-value" pairs to store data. 
# Instead of using numerical indexes (like 0, 1, 2 in Lists), 
# dictionaries use the KEYS as the index to access the values!

# --- SYNTAX EXAMPLE ---
student = {"name": "John", "age": 20}

# Accessing a value using its key (just like you would use an index)
print(student["name"]) 
# OUTPUT: John


# =====================================================================
# 🔄 CRUD OPERATIONS (Create, Read, Update, Delete)
# =====================================================================
# As the slide notes, you can perform all CRUD operations on the VALUES.
# However, you CANNOT change a KEY after it is created! 
# (To "change" a key, you must delete the old one and create a new one).

print("\n--- CRUD DEMONSTRATION ---")

# 1. CREATE (Add a brand new key-value pair)
student["grade"] = "A"
print(f"1. CREATE: {student}")
# OUTPUT: {'name': 'John', 'age': 20, 'grade': 'A'}

# 2. READ (Access an existing value)
read_value = student["age"]
print(f"2. READ: The student's age is {read_value}")
# OUTPUT: The student's age is 20

# 3. UPDATE (Change the value of an existing key)
student["age"] = 21
print(f"3. UPDATE: {student}")
# OUTPUT: {'name': 'John', 'age': 21, 'grade': 'A'}

# 4. DELETE (Remove a key-value pair entirely)
del student["grade"]
print(f"4. DELETE: {student}")
# OUTPUT: {'name': 'John', 'age': 21}
```

```python
# =====================================================================
# 🚶‍♂️ DICTIONARY TRAVERSING
# =====================================================================
# Traversing a dictionary means looping through its elements. We can 
# traverse both keys and values in a dictionary to access our data.

# =====================================================================
# 1. THE DEFAULT LOOP (Iterating over Keys)
# =====================================================================
# By default, when you loop through a dictionary, Python sets the loop 
# to iterate over the KEYS. 
# 
# However, because you have the key, you can easily access its 
# corresponding value inside the loop! This technically allows you 
# to traverse both keys and values at the exact same time.

# --- EXAMPLE FROM THE SLIDE ---
numbers = {1: 10, 2: 20, 3: 30, 4: 40}

print("--- Traversing using Keys ---")

for i in numbers:
    # 'i' represents the Key
    # 'numbers[i]' retrieves the Value for that Key
    print(i, ":", numbers[i])

# OUTPUT:
# 1 : 10
# 2 : 20
# 3 : 30
# 4 : 40


# =====================================================================
# 🌟 PRO-TIP: THE .items() METHOD
# =====================================================================
# While the method above works perfectly, Python provides a built-in 
# method called .items() that is specifically designed for traversing 
# both keys and values simultaneously. It unpacks them directly!

print("\n--- Traversing using .items() ---")

for key, value in numbers.items():
    print(f"Key: {key} | Value: {value}")

# OUTPUT:
# Key: 1 | Value: 10
# Key: 2 | Value: 20
# Key: 3 | Value: 30
# Key: 4 | Value: 40
```

```python
# =====================================================================
# 🛠️ DICTIONARY METHODS
# =====================================================================
# There are not many dictionary methods, let's see the working of some. 
# As you all know, we can use help(dict) for getting the information of all the methods available.

# --- INITIAL SETUP ---
d = {10: 100, 20: 200, 30: 300, 40: 400}
print(f"Original dictionary 'd': {d}\n")


# =====================================================================
# 1. THE .get() METHOD
# =====================================================================
# Safely retrieves the value for a specified key without crashing if 
# the key doesn't exist.

d2 = d.get(20)
print(f"Value for key 20 using .get(): {d2}\n")


# =====================================================================
# 2. THE .items() METHOD
# =====================================================================
# Retrieves all key-value pairs as a viewable object.

print("Output of d.items():")
print(d.items())
print()


# =====================================================================
# 3. THE .copy() METHOD vs. DIRECT ASSIGNMENT
# =====================================================================
# To understand why we need .copy() for dictionaries, let's look at 
# how Python handles memory references using Lists!

# ❌ THE PROBLEM WITH DIRECT ASSIGNMENT (b = a)
a = [1, 2, 3, 4, 5]
b = a
b[0] = 100

print("--- Direct Assignment (Lists) ---")
print(f"List 'a' after modifying 'b': {a}")
# Because 'b' was just a reference to 'a', changing 'b' accidentally changed 'a'!

# ✅ THE SOLUTION: USING .copy()
a = [1, 2, 3, 4, 5]
b = a.copy()
b[0] = 100

print("\n--- Using .copy() (Lists) ---")
print(f"List 'a' after modifying copied 'b': {a}")
# Now, 'a' remains safely unchanged!

# --> APPLYING THIS TO DICTIONARIES
# The exact same logic applies to dictionaries. We use .copy() to create 
# a safe, independent duplicate of the original dictionary.
d2 = d.copy()
print(f"\nSuccessfully created an independent dictionary copy: {d2}\n")


# =====================================================================
# 4. THE .clear() METHOD
# =====================================================================
# Completely empties the dictionary.

d.clear()
print(f"Dictionary 'd' after .clear(): {d}")
```

```python
#2 Write a Python program to sum all the values in a dictionary?
# =====================================================================
# ➕ SUMMING ALL VALUES IN A DICTIONARY
# =====================================================================
# Here is the logic shown in the image, along with a much faster 
# one-line shortcut!

# =====================================================================
# 1. THE LOOP METHOD (From the image)
# =====================================================================
d1 = {10: 100, 20: 200, 40: 300}

# Pro-tip: Use 'total' instead of 'sum' so we don't break Python's built-in function!
total = 0

# 'i' traverses through the keys (10, 20, 40)
for i in d1:
    # We add the value at key 'i' to our running total
    total = total + d1[i]

print(f"Total using loop: {total}")
# OUTPUT: 600


# =====================================================================
# 2. THE BUILT-IN METHOD (The Pythonic Way)
# =====================================================================
# Python has a built-in method called .values() that grabs all the 
# values from the dictionary, and a built-in sum() function to add 
# them up instantly!

d1 = {10: 100, 20: 200, 40: 300}

# All done in just one single line of code!
fast_total = sum(d1.values())

print(f"Total using sum(): {fast_total}")
# OUTPUT: 600
```

```python
#3 Count the frequency of each element in the list by using dictionary
# =====================================================================
# 📊 COUNTING ELEMENT FREQUENCY IN A LIST
# =====================================================================
# Here is the logic to count how many times each number appears in a 
# list by storing the frequencies inside a dictionary.

# =====================================================================
# 1. THE DICTIONARY LOOP METHOD (From the image)
# =====================================================================
a = [1, 1, 1, 2, 2, 2, 3, 3, 3, 4, 4, 4, 5, 5, 6, 7, 8]

d = {}

for i in a:
    # Check if the number is already a key in our dictionary
    if i in d.keys():
        d[i] += 1  # If yes, increment its existing count by 1
    else:
        d[i] = 1   # If no, add it to the dictionary with a starting count of 1

print("Frequency using dictionary logic:")
print(d)
# OUTPUT: {1: 3, 2: 3, 3: 3, 4: 3, 5: 2, 6: 1, 7: 1, 8: 1}


# =====================================================================
# 💡 MINOR OPTIMIZATION TIP
# =====================================================================
# While the logic above is perfect for learning, you don't actually 
# need to type `.keys()`. Python checks keys by default! 
# Writing `if i in d:` is actually considered more "Pythonic" and runs a tiny bit faster.


# =====================================================================
# 2. THE ADVANCED METHOD (Using collections.Counter)
# =====================================================================
# Python has a built-in library specifically designed for counting things 
# instantly. It does exactly what your loop did, but in one single line!

from collections import Counter

# Pass the list directly into Counter()
fast_counts = Counter(a)

print("\nFrequency using Counter:")
# We wrap it in dict() just to make the output look like a standard dictionary
print(dict(fast_counts)) 
# OUTPUT: {1: 3, 2: 3, 3: 3, 4: 3, 5: 2, 6: 1, 7: 1, 8: 1}
```

```python
#4 Write a Python program to combine two dictionary by adding values for common keys.
# =====================================================================
# 🤝 COMBINING DICTIONARIES & ADDING COMMON KEYS
# =====================================================================
# If two dictionaries have the exact same key, we want to add their 
# values together. Otherwise, we just add the new key-value pair.

# =====================================================================
# 1. THE LOOP METHOD (From the image)
# =====================================================================
d1 = {10: 100, 20: 200, 40: 300}
d2 = {40: 400, 50: 500, 60: 600}

# Iterate through every key in the second dictionary
for i in d2:
    # If the key already exists in d1, add d2's value to the existing value
    if i in d1.keys():
        d1[i] += d2[i]
    # If the key doesn't exist in d1, simply add it as a new entry
    else:
        d1[i] = d2[i]

print("--- Combined using loop ---")
print(d1)
# OUTPUT: {10: 100, 20: 200, 40: 700, 50: 500, 60: 600}
# Notice how the value for key 40 is now 700 (300 + 400)!

# 💡 Note: Just like before, you can simplify `if i in d1.keys():` 
# down to just `if i in d1:` for slightly cleaner and faster code!


# =====================================================================
# 2. THE ADVANCED METHOD (Using collections.Counter)
# =====================================================================
# The Counter module is actually smart enough to add values of common 
# keys together mathematically when you use the '+' operator!

from collections import Counter

d1 = {10: 100, 20: 200, 40: 300}
d2 = {40: 400, 50: 500, 60: 600}

# Convert both to Counters, add them, and convert back to a dictionary
fast_combined = dict(Counter(d1) + Counter(d2))

print("\n--- Combined using Counter ---")
print(fast_combined)
# OUTPUT: {40: 700, 10: 100, 20: 200, 50: 500, 60: 600}
```

```python
##EXCEPTION HANDLING
# 🛡️ Exception Handling: Catching Bugs Before They Crash Your Code
# ==========================================
# 2. CODE CELL (Runnable Code & Line-by-Line Comments)
# ==========================================

"""
CHILD-FRIENDLY DEFINITIONS MERGED INTO COMMENTS:
- try: Let's try riding a bike!
- except: Oops, we fell! Put on a band-aid instead of going to the hospital (crashing).
- else: We didn't fall! Let's eat some ice cream to celebrate.
- finally: Time to go home and park the bike, whether we fell or not.
- raise: A parent blowing a whistle because you rode into the street!
"""

print("--- PART 1: THE SAFETY NET (Try/Except/Else/Finally) ---")

# Step 1: We ask the user for a number.
a = int(input("tell your number :- "))

# Step 2: The 'try' block. We TRY to do something risky.
try:
    # We attempt to divide 10 by the user's number. 
    # If they typed 0, this will cause a ZeroDivisionError!
    print(10/a)

# Step 3: The 'except' block. We CATCH the mistake if it happens.
# 'Exception as err' means we catch ANY exception and give it a nickname 'err'.
except Exception as err:
    # Instead of crashing, we print a friendly message showing the error.
    print(f"sorry there is an err as {err}")

# Step 4: The 'else' block. We CELEBRATE if there were no mistakes.
else:
    # This only prints if the division in the 'try' block worked perfectly.
    print("good there is no exception")

# Step 5: The 'finally' block. We CLEAN UP, no matter what.
finally:
    # This will always print, whether we did the division or caught an error.
    print("i will run no matter what")

# A normal print statement just to show the program is still alive!
print("ok i have done the division")


print("\n--- PART 2: YELLING 'TIMEOUT' (Raising Exceptions) ---")

# We ask the user for their age.
age = int(input("tell your age :- "))

try:
    # We create a strict rule: Age must be between 10 and 18.
    if age < 10 or age > 18:
        # If they break the rule, we MANUALLY throw an error using 'raise'.
        # We are creating a ValueError on purpose.
        raise ValueError("your age must be between 10 and 18")
    else:
        # If they follow the rule, they get in!
        print("welcome to the club")

except Exception as err:
    # If our 'raise' triggered, Python jumps down here and catches it.
    print(f"an error occured as {err}")

print("the club will start soon")


# ==========================================
# 3. PROGRAM FLOW & DRY RUN (The Execution Trace)
# ==========================================
"""
DRY RUN SCENARIO 1 (Happy Path - User enters 2):
1. `a = 2`
2. Enter `try`. Python calculates `10 / 2`.
3. Output: 5.0
4. Skip `except` (no errors!).
5. Go to `else`. Output: "good there is no exception"
6. Go to `finally`. Output: "i will run no matter what"
7. Output: "ok i have done the division"

DRY RUN SCENARIO 2 (Uh Oh Path - User enters 0):
1. `a = 0`
2. Enter `try`. Python tries `10 / 0`. 
3. BOOM! Division by zero error! Python instantly leaves the `try` block.
4. Go to `except`. Python catches the ZeroDivisionError as `err`.
5. Output: "sorry there is an err as division by zero"
6. Skip `else` (because we had an error).
7. Go to `finally`. Output: "i will run no matter what"
8. Output: "ok i have done the division" (The program survived!)
"""

# ==========================================
# 4. THE "PYTHONIC" PRO-TIP (Bonus)
# ==========================================
"""
PRO-TIP: Catching "Exception" is like using a giant fishing net to catch a 
specific tiny fish; you might accidentally catch a boot! 

Professional Python developers specify EXACTLY which error they expect so they 
don't accidentally hide other bugs. 

PYTHONIC WAY:
try:
    a = int(input("tell your number :- "))
    print(10 / a)
except ZeroDivisionError:
    print("Oops! You can't divide by zero. Try another number.")
except ValueError:
    print("Oops! That wasn't a number at all. Please type digits.")
"""
```

```python
#FILE HANDLING
# ==========================================
# 2. CODE CELL (Runnable Code & Line-by-Line Comments)
# ==========================================

"""
CHILD-FRIENDLY EXPLANATION (MERGED):
- open(): Opening the backpack.
- read(): Looking at everything inside.
- write(): Putting new drawings inside.
- close(): Zipping the backpack shut!
"""

import os # We import this just to clean up our playground later!

print("--- 1. WRITING TO A FILE ('w' mode) ---")
# We open 'superman.txt' in 'w' (Write) mode. 
# If it doesn't exist, Python creates it for us.
r = open("superman.txt", 'w')
# We write a sentence inside. This acts like a giant eraser first, then writes.
r.write("Hello this akarsh and I am writing inside this file ")
# Always remember to zip the backpack shut!
r.close()
print("Successfully wrote to superman.txt!")


print("\n--- 2. APPENDING TO A FILE ('a' mode) ---")
# We open the same file, but this time in 'a' (Append) mode.
# This means we add to the end, without erasing the old stuff.
r = open("superman.txt", 'a')
# We add a new sentence to the very end of the file.
r.write("and now I am appending some content inside the file ")
# Zip it shut again!
r.close()
print("Successfully appended to superman.txt!")


print("\n--- 3. READING A FILE ('r' mode) ---")
# Now we want to read it. By default, open() uses 'r' mode if we don't say anything.
# We can also use a raw string (putting an 'r' before the quotes) if we have a big, messy computer path 
# like r'/Users/akarshvyas/Desktop/Elevate40.txt'.
p = open('superman.txt') # 'superman.txt' is in our current folder.
# p.read() reads EVERYTHING inside the file.
print("Here is what is in the file:")
print(p.read())
p.close()


print("\n--- 4. THE MAGIC KEYWORD ('with') ---")
# Sometimes we forget to close files. Python has a magic word 'with'.
# 'with' acts like a helpful robot parent that automatically zips the backpack shut for us when we are done!
# We open a pretend file 'data.txt' (let's use our superman file instead so it runs).
with open("superman.txt", "r") as f:
    content = f.read()
    print("Reading using 'with':")
    print(content)
# Notice there is no f.close() here! The 'with' block handled it automatically.


# Playground Cleanup (so we don't leave messy files on your computer)
os.remove("superman.txt")


# ==========================================
# 3. PROGRAM FLOW & DRY RUN (The Execution Trace)
# ==========================================
"""
EXECUTION TRACE:
1. `r = open("superman.txt", 'w')`: Python checks the folder. No file exists. It creates it.
2. `r.write(...)`: Python writes "Hello this akarsh..." into memory/buffer.
3. `r.close()`: Python saves the buffer to the hard drive and locks the file.

4. `r = open("superman.txt", 'a')`: Python finds the file and puts the pen at the very end of the text.
5. `r.write(...)`: Python writes "and now I am appending..." right next to the previous sentence.
6. `r.close()`: Saves and locks.

7. `p = open('superman.txt')`: Python opens the file in read-only mode.
8. `p.read()`: Python sucks up all the text and prints: 
   "Hello this akarsh and I am writing inside this file and now I am appending some content inside the file "
9. `p.close()`: File is closed.

10. `with open(...)`: Python opens it again. 
11. `f.read()` reads the content.
12. Once the indented block finishes, Python's internal logic calls `f.close()` silently!
"""


# ==========================================
# 4. THE "PYTHONIC" PRO-TIP (Bonus)
# ==========================================
"""
PRO-TIP: 
While writing `file = open(...)` and `file.close()` works, a professional 
developer almost NEVER writes file handling this way. If an error happens 
while the file is open, the program crashes and `.close()` is never reached, 
leaving the file locked!

PYTHONIC WAY: Always use the Context Manager (`with` statement).
It guarantees the file is closed no matter what, even if a giant error happens!

with open("my_important_file.txt", "w") as file:
    file.write("Safe and sound!")
    # The file automatically closes right here!
"""
```

```python
# FILE HANDLING PROJECT IN PYTHON

# ==========================================
# 2. CODE CELL (Runnable Code & Line-by-Line Comments)
# ==========================================

from pathlib import Path # Brings in the 'Path' tool to handle file locations like a GPS
import os # Brings in the 'os' module to interact with the computer's operating system

def readfielandfolder(): # Defines a new reusable block of code (function) to list files
    path = Path('') # Creates a Path object pointing to the current working folder (empty string)
    items = list(path.rglob('*')) # Searches recursively for everything ('*') and converts the result into a list
    for i, items in enumerate(items): # Loops through the items, using 'enumerate' to get an automatic counter 'i'
        print(f"{i+1}  : {items} ") # Prints the counter (adding 1 so it starts at 1, not 0) and the file name


def createfile(): # Defines a function to create a brand new file
    try: # Starts a block to test for errors during execution
        readfielandfolder() # Calls our previously defined function to show the user existing files
        name = input("please tell your file name :- ") # Asks the user for a name and stores it in the variable 'name'
        p = Path(name) # Converts the string name into a proper Path object
        
        if not p.exists(): # Checks the condition: if this file DOES NOT already exist on the computer
            with open(p, "w") as fs: # Opens the file safely in 'w' (write) mode, assigning it to variable 'fs'
                data = input("what you want to write in this file :- ") # Prompts user for text and stores it in 'data'
                fs.write(data) # Writes the stored text data directly into the open file
            
            print(f"FILE CREATED SUCCESSFULLY") # Prints a success message to the screen
        else: # If the 'if' condition is false (meaning the file DOES exist), run this block instead
            print("this file already exist") # Warns the user they can't create it again
        
    except Exception as err: # Catches any unexpected errors that happened in the 'try' block
        print(f"An error occured as {err}") # Prints the specific error message


def readfile(): # Defines a function to open and read a file's contents
    try: # Starts the error-monitoring block
        readfielandfolder() # Shows the user what files are available
        name = input("which file you want to read ") # Gets the target file name from the user
        p = Path(name) # Converts the input string into a Path object
        
        if p.exists() and p.is_file(): # Checks TWO conditions: does it exist? AND is it actually a file (not a folder)?
            with open(p, 'r') as fs: # Opens the file safely in 'r' (read) mode
                data = fs.read() # Reads the entire contents of the file and stores it in 'data'
                print(data) # Prints the contents to the screen
            
            print("Readed successfully ") # Prints a confirmation message
        else: # Executes if the file doesn't exist or is a folder
            print("the file doesnot exist") # Warns the user
            
    except Exception as err: # Catches any crashes (like permission issues)
        print(f"An error occured as {err}") # Displays the error


def updatefile(): # Defines a function to modify an existing file
    try: # Starts the error-monitoring block
        readfielandfolder() # Shows available files
        name = input("tell which file you want to update :- ") # Asks the user which file to target
        p = Path(name) # Creates a Path object for the target file
        
        if p.exists() and p.is_file(): # Checks if the file is valid and exists
            print("press 1 for changing the name of your file :- ") # Prints a menu option
            print("press 2 for overwriting the data of your file ") # Prints a menu option
            print("press 3 for appending some content in your file ") # Prints a menu option

            res = int(input("tell your response :- ")) # Gets user input and converts it from text to a whole number (integer)

            if res == 1: # Checks if the user chose option 1
                name2 = input("tell  your new file name :- ") # Asks for the new file name
                p2 = Path(name2) # Creates a Path object for the NEW name
                p.rename(p2) # Uses the pathlib tool to easily rename the file from 'p' to 'p2'
            
            if res == 2: # Checks if the user chose option 2
                with open(p, 'w') as fs: # Opens the file in 'w' (write/overwrite) mode
                    data = input("tell what you want to write this is overwrite the data :- ") # Gets new text
                    fs.write(data) # Writes the text, completely erasing the old contents
        
            if res == 3: # Checks if the user chose option 3
                with open(p, 'a') as fs: # Opens the file in 'a' (append) mode
                    data = input("tell what you want to append :- ") # Gets the text to add
                    fs.write(" " + data) # Adds a space and the new text to the very end of the file

    except Exception as err: # Catches any errors
        print(f"an error occured as {err}") # Prints the error message


def deletefile(): # Defines a function to delete a file
    try: # Starts error-monitoring block
        readfielandfolder() # Lists current files
        name = input("whuich file you want to delete :- ") # Asks user which file to destroy
        p = Path(name) # Converts input to Path object

        if p.exists() and p.is_file(): # Checks if the file is real and is not a directory
            os.remove(name) # Uses the 'os' module to permanently delete the file from the hard drive
            print("file removes suiccessfully ") # Confirms deletion
        else: # Executes if file isn't found
            print("No such file exist") # Warns user
    
    except Exception as err: # Catches errors
        print(f"An error occured as {err}") # Prints error


# --- MAIN PROGRAM EXECUTION STARTS HERE ---

print("press 1 for creating a file") # Displays main menu option 1
print("press 2 for reading a file") # Displays main menu option 2
print("press 3 for updating a file") # Displays main menu option 3
print("press 4 for deletion a file") # Displays main menu option 4

check = int(input("please tell your response :- ")) # Takes user input, converts to integer, stores in 'check'

if check == 1: # Evaluates if user entered 1
    createfile() # Calls the createfile function

if check == 2: # Evaluates if user entered 2
    readfile() # Calls the readfile function

if check == 3: # Evaluates if user entered 3
    updatefile() # Calls the updatefile function

if check == 4: # Evaluates if user entered 4
    deletefile() # Calls the deletefile function


# ==========================================
# 3. PROGRAM FLOW & DRY RUN (The Execution Trace)
# ==========================================
"""
EXECUTION TRACE (Scenario: Updating a file by appending text):
1. Python reads all 'def' blocks into memory without running them yet.
2. Reaches the Main Menu `print` statements and displays them.
3. Reaches `check = int(input(...))`. User types '3' and presses Enter.
4. `check` becomes the integer 3.
5. Python evaluates `if check == 1:` -> False.
6. Python evaluates `if check == 2:` -> False.
7. Python evaluates `if check == 3:` -> True! Execution jumps into `updatefile()`.
8. Inside `updatefile()`:
   a. Enters `try` block.
   b. Calls `readfielandfolder()` to display directory contents.
   c. Prompts for filename. User types "log.txt". `p = Path("log.txt")`.
   d. `if p.exists() and p.is_file():` evaluates to True.
   e. Displays the update sub-menu (1=rename, 2=overwrite, 3=append).
   f. `res = int(input(...))` -> User types '3'.
   g. `if res == 1:` -> False.
   h. `if res == 2:` -> False.
   i. `if res == 3:` -> True! 
   j. Opens "log.txt" in 'a' (append) mode using `with open(...)`.
   k. Prompts for append data. User types "System checked."
   l. Executes `fs.write(" System checked.")`.
   m. File closes automatically.
9. Function finishes, program terminates.
"""

# ==========================================
# 4. THE "PYTHONIC" PRO-TIP (Bonus)
# ==========================================
"""
While it is great to understand what every statement does, writing a comment on 
EVERY single line is considered a "code smell" in professional development because 
it makes the code visually cluttered. 

PYTHONIC WAY:
Write code that explains itself using good variable names, and use "Docstrings" 
to explain the PURPOSE of a function, rather than explaining the syntax line-by-line.

def deletefile(file_name: str) -> None:
    \"\"\"
    Attempts to delete a specified file from the current directory.
    Alerts the user if the file does not exist or cannot be accessed.
    \"\"\"
    target_path = Path(file_name)
    if target_path.is_file():
        target_path.unlink()  # pathlib's built-in delete method
        print(f"{file_name} deleted.")
"""
```
