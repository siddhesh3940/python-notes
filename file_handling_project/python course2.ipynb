```python
#5 accept a year and check if it is a leap year
year = int(input("give me a year-"))
if year%4 == 0:  #if year devides by 4 with 0 remainder then it is a leap year
    print(f"{year} is a leap year")
elif year%100 == 0:
    print(f"{year} is not a leap year")
elif year%400 == 0:
    print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
```
